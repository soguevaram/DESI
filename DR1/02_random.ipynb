```python
def cartesianas(data):
    # Calcular distancia comóvil
    data['distance'] = Planck18.comoving_distance(data['Z'])

    ra_values = np.array(data['RA'])
    dec_values = np.array(data['DEC'])

    coords = SkyCoord(
        ra = ra_values * u.deg,
        dec = dec_values * u.deg,
        distance = data['distance'],
        frame = 'icrs'
    )
    data['x'] = coords.cartesian.x.value
    data['y'] = coords.cartesian.y.value
    data['z'] = coords.cartesian.z.value
```

```python
  def url(rand_generator):

      urls = [f"https://data.desi.lbl.gov/public/dr1/survey/catalogs/dr1/LSS/iron/LSScats/v1.5/LRG_NGC_{i}_clustering.ran.fits"
              for i in range(18)]

      url = rand_generator.choice(urls)
      rand = Table.read(url)
      return rand

  def save_file(data, number, n_random=100, seed=42):
      rng = np.random.default_rng(seed)
      Path("rand").mkdir(exist_ok=True)

      for i in range(n_random):
          print(f'{i}/{n_random}')
          rand = url(rng)

          if number == 1:
              mask = (rand['RA'] >= 110) & (rand['RA'] <= 260) & (rand['DEC'] >= -10) & (rand['DEC'] <= 8)
          elif number == 2:
              mask = (rand['RA'] >= 180) & (rand['RA'] <= 260) & (rand['DEC'] >= 30) & (rand['DEC'] <= 40)
          else:
              raise ValueError("Orientation must be 1 or 2.")

          rand = rand[mask]

          idx_random = rng.choice(len(rand), size=len(data), replace=False)
          rand_subset = rand[idx_random]

          cartesianas(rand_subset)
          rand_subset = rand_subset[['TARGETID','x','y','z']]
          rand_subset['type'] = 'rand'

          filename_rand = f"/content/drive/MyDrive/DESI/rand/LRG_NGC_{number}_random_{i}_filt.fits"

          rand_subset.write(filename_rand, overwrite=True)

          del rand_subset
          gc.collect()

      print("Saved files!")
```

```python
%%time
save_file(filt_n1,1)
```
